```python
import numpy as np
import random
import copy

np.random.seed(42)
PRICE_DATA = 100.0 + np.cumsum(np.random.normal(0.15, 1.5, 365))
START_CAPITAL = 100000.00

def calculate_moving_average(prices, window_size):
    ma_array = np.zeros(len(prices))
    for i in range(len(prices)):
        if i < window_size:
            ma_array[i] = prices[i]
        else:
            ma_array[i] = np.mean(prices[i - window_size + 1 : i + 1])
    return ma_array

POPULATION_SIZE = 50
MUTATION_RATE   = 0.10
CROSSOVER_RATE  = 0.80
MAX_GENERATIONS = 30

def create_individual():
    while True:
        X = random.randint(5, 50)
        Y = random.randint(20, 200)
        if X < Y:
            return [X, Y]

POPULATION = [create_individual() for _ in range(POPULATION_SIZE)]

def evaluate_fitness(individual):
    X = individual[0]
    Y = individual[1]

    fast_ma = calculate_moving_average(PRICE_DATA, X)
    slow_ma = calculate_moving_average(PRICE_DATA, Y)

    current_cash = START_CAPITAL
    shares_held  = 0

    for t in range(1, len(PRICE_DATA)):
        current_price = PRICE_DATA[t]

        if shares_held == 0 and fast_ma[t] > slow_ma[t] and fast_ma[t-1] <= slow_ma[t-1]:
            shares_held  = current_cash / current_price
            current_cash = 0.0

        elif shares_held > 0 and fast_ma[t] < slow_ma[t] and fast_ma[t-1] >= slow_ma[t-1]:
            current_cash = shares_held * current_price
            shares_held  = 0

    if shares_held > 0:
        current_cash = shares_held * PRICE_DATA[-1]

    net_profit = current_cash - START_CAPITAL
    return max(0.0, net_profit)

def select_parent(population, fitness_scores):
    index_1 = random.randint(0, POPULATION_SIZE - 1)
    index_2 = random.randint(0, POPULATION_SIZE - 1)

    if fitness_scores[index_1] > fitness_scores[index_2]:
        return copy.deepcopy(population[index_1])
    else:
        return copy.deepcopy(population[index_2])

def perform_crossover(parent_1, parent_2):
    if random.random() < CROSSOVER_RATE:
        child_1 = [parent_1[0], parent_2[1]]
        child_2 = [parent_2[0], parent_1[1]]

        if child_1[0] >= child_1[1]: child_1[1] = child_1[0] + 15
        if child_2[0] >= child_2[1]: child_2[1] = child_2[0] + 15
        return [child_1, child_2]
    else:
        return [copy.deepcopy(parent_1), copy.deepcopy(parent_2)]

def apply_mutation(individual):
    if random.random() < MUTATION_RATE:
        individual[0] = individual[0] + random.choice([-2, -1, 1, 2])
        if individual[0] < 5:  individual[0] = 5
        if individual[0] > 50: individual[0] = 50

    if random.random() < MUTATION_RATE:
        individual[1] = individual[1] + random.choice([-5, -3, 3, 5])
        if individual[1] < 20:  individual[1] = 20
        if individual[1] > 200: individual[1] = 200

    if individual[0] >= individual[1]:
        individual[1] = individual[0] + 5

    return individual

GLOBAL_BEST_STRATEGY = None
GLOBAL_BEST_FITNESS  = -1.0

print("Initializing Genetic Algorithm Optimization Run...\n")

for generation in range(1, MAX_GENERATIONS + 1):

    fitness_scores = [evaluate_fitness(ind) for ind in POPULATION]

    best_idx = np.argmax(fitness_scores)
    if fitness_scores[best_idx] > GLOBAL_BEST_FITNESS:
        GLOBAL_BEST_FITNESS = fitness_scores[best_idx]
        GLOBAL_BEST_STRATEGY = copy.deepcopy(POPULATION[best_idx])

    print(f"Generation: {generation:02d} | Current Strategy Window: {POPULATION[best_idx]} | Top Net Profit: Rs.{GLOBAL_BEST_FITNESS:,.2f}")

    next_generation_pool = []
    while len(next_generation_pool) < POPULATION_SIZE:
        p1 = select_parent(POPULATION, fitness_scores)
        p2 = select_parent(POPULATION, fitness_scores)

        children = perform_crossover(p1, p2)

        c1 = apply_mutation(children[0])
        c2 = apply_mutation(children[1])

        next_generation_pool.append(c1)
        next_generation_pool.append(c2)

    POPULATION = next_generation_pool[:POPULATION_SIZE]

print("\n=======================================================")
print("          GA ALGORITHM CONVERGENCE SUCCESSFUL          ")
print("=======================================================")
print(f"Optimal Fast Moving Average (X): {GLOBAL_BEST_STRATEGY[0]} days")
print(f"Optimal Slow Moving Average (Y): {GLOBAL_BEST_STRATEGY[1]} days")
print(f"Maximum Achieved Backtest Profit: Rs.{GLOBAL_BEST_FITNESS:,.2f}")
print("=======================================================")
```
