```python
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import cross_val_score
from sklearn.ensemble import RandomForestClassifier

data = load_breast_cancer()
X, y = data.data, data.target
num_particles = 20
max_iter = 20
num_features = X.shape[1]

w = 0.7
c1 = 1.5
c2 = 1.5

X_pos = np.random.randint(2, size=(num_particles, num_features))
V_vel = np.random.uniform(-4, 4, size=(num_particles, num_features))

Pbest_pos = np.copy(X_pos)
Pbest_fitness = np.zeros(num_particles)
Gbest_pos = np.zeros(num_features)
Gbest_fitness = 0.0

def calculate_fitness(feature_mask):
    if np.sum(feature_mask) == 0:
        return 0.0
    X_subset = X[:, feature_mask == 1]
    model = RandomForestClassifier(n_estimators=10, random_state=42)
    scores = cross_val_score(model, X_subset, y, cv=3, scoring='accuracy')
    return scores.mean()

for i in range(num_particles):
    score = calculate_fitness(X_pos[i])
    Pbest_fitness[i] = score
    if score > Gbest_fitness:
        Gbest_fitness = score
        Gbest_pos = np.copy(X_pos[i])

print("Starting Binary PSO Feature Selection...")
for it in range(max_iter):
    for i in range(num_particles):
        r1, r2 = np.random.rand(num_features), np.random.rand(num_features)

        V_vel[i] = (w * V_vel[i] +
                    c1 * r1 * (Pbest_pos[i] - X_pos[i]) +
                    c2 * r2 * (Gbest_pos - X_pos[i]))

        sigmoid = 1 / (1 + np.exp(-V_vel[i]))
        X_pos[i] = (np.random.rand(num_features) < sigmoid).astype(int)

        current_fitness = calculate_fitness(X_pos[i])

        if current_fitness > Pbest_fitness[i]:
            Pbest_fitness[i] = current_fitness
            Pbest_pos[i] = np.copy(X_pos[i])

            if current_fitness > Gbest_fitness:
                Gbest_fitness = current_fitness
                Gbest_pos = np.copy(X_pos[i])

    print(f"Iteration {it+1}/{max_iter} | Best Cross-Val Accuracy: {Gbest_fitness * 100:.2f}%")

selected_indices = np.where(Gbest_pos == 1)[0]
selected_names = [data.feature_names[idx] for idx in selected_indices]

print("\n=== FINAL RESULTS ===")
print(f"Total features originally: {num_features}")
print(f"Features selected by swarm: {len(selected_indices)}")
print(f"Selected Feature Names: {selected_names}")
print(f"Optimized ML Model Accuracy: {Gbest_fitness * 100:.2f}%")
```
